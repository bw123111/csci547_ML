````python
# Logistic regression class
class Logistic_regression:
    # Class constructor
    def __init__(self):
        self.w = None     # logistic regression weights
        self.saved_w = [] # Since this is a small problem, we can save the weights
                          #  at each iteration of gradient descent to build our
                          #  learning curves
        self.saved_train_costs = []  # Store training costs computed during fit
        self.saved_test_costs = []   # Store test costs computed during fit
        # returns nothing
        pass

    # Method for calculating the sigmoid function of w^T X for an input set of weights
    def sigmoid(self, X, w):
      z = np.dot(X, w)
      z = np.clip(z, -500, 500)  # Prevent extreme values, had a Runtime error before...
      return 1/(1+np.exp(-z))
        # returns the value of the sigmoid

    # Cost function for an input set of weights
    def cost(self, X, y, w):
        # Add small epsilon to prevent log(0) (undefined)
        epsilon = 1e-15
        y_pred = self.sigmoid(X, w)
        y_pred = np.clip(y_pred, epsilon, 1 - epsilon)

        return (-np.mean(y*np.log(y_pred) + (1-y)*np.log(1-y_pred)))
        # this is the coded version of the gradient descent cost function from lecture.
        # y hat is the prediction, i.e. the sigmoid function.
        # returns the average cross entropy cost

    # Update the weights in an iteration of gradient descent, return difference in weights
    def gradient_descent(self, X, y, lr):
        n_samples = X.shape[0]

        # calc predictions
        y_pred = self.sigmoid(X, self.w)

        # calc gradient
        gradient = (1/n_samples) * np.dot(X.T, (y_pred - y))

        # Store old weights to calc change
        old_w = self.w.copy()

        # Update weights
        self.w = self.w - lr * gradient

        # calc magnitude of change
        weight_change = np.linalg.norm(self.w - old_w)

        # returns a scalar of the magnitude of the Euclidean norm
        #  of the change in the weights during one gradient descent step
        return weight_change

    # Fit the logistic regression model to the data through gradient descent
    def fit(self, X, y, w_init, lr, X_test=None, y_test=None, delta_thresh=1e-6, max_iter=5000, verbose=False):
        # Note the verbose flag enables you to print out the weights at each iteration
        #  (optional - but may help with one of the questions)

        # Initialize weights and clear saved costs
        self.w = w_init.copy()
        self.saved_w = [self.w.copy()]  # Save initial weights
        self.saved_train_costs = [self.cost(X, y, self.w)]  # Save initial train cost

        # Save initial test cost if test data provided
        if X_test is not None and y_test is not None:
            self.saved_test_costs = [self.cost(X_test, y_test, self.w)]
        else:
            self.saved_test_costs = []

        # Gradient descent loop
        for iteration in range(max_iter):
            # Perform one step of gradient descent
            weight_change = self.gradient_descent(X, y, lr)

            # Save weights and costs at each iteration
            self.saved_w.append(self.w.copy())
            self.saved_train_costs.append(self.cost(X, y, self.w))

            # Save test cost if test data provided
            if X_test is not None and y_test is not None:
                self.saved_test_costs.append(self.cost(X_test, y_test, self.w))

            if verbose:
                print(f"Iteration {iteration}: Cost = {self.saved_train_costs[-1]:.6f}, Weight change = {weight_change:.8f}")

            # Stopping criterion: check if change in weights is small enough
            if weight_change < delta_thresh:
                if verbose:
                    print(f"Converged at iteration {iteration}")
                break

        # returns nothing
        pass

    # Use the trained model to predict the confidence scores (prob of positive class in this case)
    def predict_proba(self, X):
        # returns the confidence score for the each sample
        return self.sigmoid(X, self.w)

    # Use the trained model to make binary predictions
    def predict(self, X, thresh=0.5):
        # returns a binary pred per sample based on probability pred
        probas = self.predict_proba(X)
        return np.array([0 if p <= thresh else 1 for p in probas])

    # Stores the learning curves from saved weights from gradient descent
    def learning_curve(self, X=None, y=None):

        # If only X and y, calc costs from saved weights (slow!)
        if X is not None and y is not None:
            return [self.cost(X, y, w) for w in self.saved_w]

        # else  return saved costs (fast!!)
        # returns the value of the cost function from each step in gradient descent
        # from the last model fitting run
        return self.saved_train_costs, self.saved_test_costs

    # Appends a column of ones as the first feature to account for the bias term
    def prepare_x(self, X):
        # returns the X with a new feature of all ones (a column that is the new column 0)
        n_samples = X.shape[0]
        ones = np.ones((n_samples, 1))
        return np.hstack([ones, X])
````

````python
learning_rates = [1.0, 1e-2, 1e-4, 1e-6]

plt.figure(figsize=(14, 8))

# Dict to store results
results = {}

# Train model for each learning rate
for lr in learning_rates:
    print(f"Training with learning rate: {lr}")

    model = Logistic_regression()
    X_train_prep = model.prepare_x(X_train)
    X_test_prep = model.prepare_x(X_test)

    # Initialize weights randomly between 0 and 1 according to instructions
    np.random.seed(42)  # For reproducibility across runs
    w_init = np.random.rand(X_train_prep.shape[1])

    # Fit the model - add test data to compute costs during training and testing
    model.fit(X_train_prep, y_train, w_init, lr=lr,
              X_test=X_test_prep, y_test=y_test,
              delta_thresh=1e-6, max_iter=5000, verbose=False)

    # Get learning curves
    train_costs, test_costs = model.learning_curve()

    # Store results
    results[lr] = {
        'train_costs': train_costs,
        'test_costs': test_costs,
        'iterations': len(train_costs)
    }

    # Plot training curve
    plt.plot(range(len(train_costs)), train_costs,
             label=f'Train (lr={lr})', linewidth=2, linestyle='-')

    # Plot test curve
    plt.plot(range(len(test_costs)), test_costs,
             label=f'Test (lr={lr})', linewidth=2, linestyle='--')

# Customize plot
plt.xlabel('Iteration', fontsize=12)
plt.ylabel('Average Cost', fontsize=12)
plt.title('Learning Curves: Training and Test Cost vs Iteration\nfor Different Learning Rates',
          fontsize=14, fontweight='bold')
plt.legend(loc='best', fontsize=10)
plt.grid(True, alpha=0.3)
plt.tight_layout()

# Optional: Use log scale for x-axis if curves converge quickly
# plt.xscale('log')

plt.show()

# Print summary comparison (used Claude AI to generate the nice tabular spacing/format)
print("SUMMARY")
print(f"{'Learning Rate':<15} {'Iterations':<12} {'Final Train Cost':<18} {'Final Test Cost':<18}")

for lr in learning_rates:
    iterations = results[lr]['iterations'] - 1
    final_train = results[lr]['train_costs'][-1]
    final_test = results[lr]['test_costs'][-1]
    print(f"{lr:<15.0e} {iterations:<12} {final_train:<18.6f} {final_test:<18.6f}")
````
